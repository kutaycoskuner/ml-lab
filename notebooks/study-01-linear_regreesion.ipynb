# Linear Regression (ai-generated)

In linear regression, we have a set of data points in a 2D coordinate system:

$$
(x_1,y_1), (x_2,y_2),\ldots,(x_n,y_n)
$$

We want to find a straight line that represents the data as well as possible.

The line is described by:

$$
\hat{y}=wx+b
$$

where:

- $w$ = **slope**
- $b$ = **intercept**
- $\hat{y}$ = **predicted value**

For each data point, we calculate the difference between the actual value $y_i$ and the predicted value $\hat{y}_i$:

$$
e_i=y_i-\hat{y}_i
$$

This difference is called the **residual**.

The goal is **not simply to minimize the difference for one point**. Instead, we want to find the line that minimizes the overall error across **all data points**.

A common approach is **Ordinary Least Squares (OLS)**, where we minimize the sum of squared residuals:

$$
J(w,b)=\sum_{i=1}^{n}(y_i-(wx_i+b))^2
$$

This function $J(w,b)$ is called the **cost function**.

Therefore, the basic idea of linear regression is:

$$
\boxed{\text{Find } w,b \text{ such that } J(w,b) \text{ is minimized}}
$$

## Important Concepts

To understand linear regression, we need the following concepts:

1. **Data points** — $(x_i,y_i)$
2. **Linear model** — $\hat{y}=wx+b$
3. **Prediction** — $\hat{y}_i$
4. **Residual / error** — $y_i-\hat{y}_i$
5. **Loss / cost function** — measures the total error
6. **Least squares** — minimizes the squared errors
7. **Optimization** — finds the $w,b$ that minimize the cost function
8. **Derivative / gradient** — tells us how the cost changes when $w$ or $b$ changes
9. **Gradient descent** — an iterative method for finding the minimum

## Key Idea

> **Linear regression finds the parameters of a line that minimize the overall prediction error, commonly measured as the sum of squared residuals.**

# Step 1: Generate synthetic data and visualize

In order to deal with some data, we need some data. For this task, I will generate some data and somehow generate it in a way that it could somehow resembles a line.

Then we need to visualize data so we can clearly understand how it looks like.

```python
# we import those libraries, as we need to generate as well as visualize
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Generate 50 evenly spaced values between 1 and 10
study_hours = np.linspace(1, 10, 50)

# Generate exam scores based on a linear relationship
# np.random.normal(mean, standard_deviation, number_of_values)
# adds normally distributed random noise to make the data more realistic
exam_scores = 5 * study_hours + 30 + np.random.normal(0, 5, 50)

# try: try to change standard deviation to more scattered results.

# Create a scatter plot where each point represents one observation
plt.scatter(study_hours, exam_scores)

# Label the x-axis
plt.xlabel("Study Hours")

# Label the y-axis
plt.ylabel("Exam Score")

# Add a title to the plot
plt.title("Study Hours vs. Exam Score")

# Display the plot
plt.show()
```

# Step 2: Understand the line parameters

$$
\hat{y}=wx+b
$$

lets understand this first. we have $x$ as variable. $w$ as coefficient of variable as well as, $b$ as constant. In this setup, $w$ will control the slope of line, whereas b will raise or reduce it. Lets try by changing parameters of the line on below code block.

```python
# Define the parameters of the line
w = 0
b = 0

# Create x-values for the line
x_line = np.linspace(1, 10, 100)

# Calculate y-values using y = wx + b
y_line = w * x_line + b

# Plot the data and the line
plt.scatter(study_hours, exam_scores, label="Data")
plt.plot(x_line, y_line, label=f"y = {w}x + {b}")

plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Effect of w and b on a Line")
plt.legend()
plt.show()
```

# Step 3: Guess a line (optional but good for comprehension)

# Step 4: Understanding the Concept of Mean Square Error (MSE)

We have a line that is supposed to represent our data.
Now we want to answer: How different is this line from the actual data?

For each data point, we can look at the vertical distance between the actual point and the line.

The line gives us a predicted value: $\hat{y}=wx+b$ for x.

For a data point with actual value $y$, the difference is: $e=y−\hat{y}$​   
We call this difference the error. How our estimation different from the original data.

But we dont have one data, we have multiple. If we sum up all errors we can get a single number precisely representing the error. So we can compare more lines if the one is better than other!

But there is some problem, some errors are positive whereas others are negative, summing them will not give a good result. We eventually interested with the distance not the positive or negative. So overcoming this problem we square the errors so we always have positive outcomes. 

Finally we average the sum of errors which gives us the Mean Squared Error (MSE).

$$
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

Smaller MSE means that line is closer to overall data,
Larger MSE means that line is farther from the overal data.

Now lets practice this.





```python
# Calculate predictions for the actual data points
y_pred = w * study_hours + b

# Calculate the error for each data point
errors = exam_scores - y_pred

# Square the errors
squared_errors = errors ** 2

# Calculate the Mean Squared Error
mse = np.mean(squared_errors)

print("Errors:", errors)
print("Squared errors:", squared_errors)
print("MSE:", mse)
```

# Step 5: Comparing the lines

Now we have a numeric value that clearly tells us if a line is better than other for representing this particular data thanks to mathematics. With this knowledge, we can now compare various lines and see the results. Lets do that.

```python
# Define three different lines
w1, b1 = 4, 35
w2, b2 = 5, 30
w3, b3 = 6, 25

# Calculate predictions
y_pred_1 = w1 * study_hours + b1
y_pred_2 = w2 * study_hours + b2
y_pred_3 = w3 * study_hours + b3

# Calculate MSE for each line
mse_1 = np.mean((exam_scores - y_pred_1) ** 2)
mse_2 = np.mean((exam_scores - y_pred_2) ** 2)
mse_3 = np.mean((exam_scores - y_pred_3) ** 2)

print("Line 1: w =", w1, "b =", b1, "MSE =", mse_1)
print("Line 2: w =", w2, "b =", b2, "MSE =", mse_2)
print("Line 3: w =", w3, "b =", b3, "MSE =", mse_3)
```

among these lines clearly Line 2 appears as a best candidate to represent our data with the value of 22.6...

Lets check the graph with actual data.

```python
# Create x-values for the lines
x_line = np.linspace(1, 10, 100)

# Calculate y-values
y_line_1 = w1 * x_line + b1
y_line_2 = w2 * x_line + b2
y_line_3 = w3 * x_line + b3

# Plot the data
plt.scatter(study_hours, exam_scores, label="Data")

# Plot the three lines
plt.plot(x_line, y_line_1, label=f"Line 1: y = {w1}x + {b1}")
plt.plot(x_line, y_line_2, label=f"Line 2: y = {w2}x + {b2}")
plt.plot(x_line, y_line_3, label=f"Line 3: y = {w3}x + {b3}")

plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Comparing Different Lines")
plt.legend()
plt.show()
```

As graph by the graph as well we can clearly see that Line 2 providing the best result on representing this data set.
