# Linear Regression (ai-generated)

In linear regression, we have a set of data points in a 2D coordinate system:

$$
(x_1,y_1), (x_2,y_2),\ldots,(x_n,y_n)
$$

We want to find a straight line that represents the data as well as possible.

The line is described by:

$$
\hat{y}=wx+b
$$

where:

- $w$ = **slope**
- $b$ = **intercept**
- $\hat{y}$ = **predicted value**

For each data point, we calculate the difference between the actual value $y_i$ and the predicted value $\hat{y}_i$:

$$
e_i=y_i-\hat{y}_i
$$

This difference is called the **residual**.

The goal is **not simply to minimize the difference for one point**. Instead, we want to find the line that minimizes the overall error across **all data points**.

A common approach is **Ordinary Least Squares (OLS)**, where we minimize the sum of squared residuals:

$$
J(w,b)=\sum_{i=1}^{n}(y_i-(wx_i+b))^2
$$

This function $J(w,b)$ is called the **cost function**.

Therefore, the basic idea of linear regression is:

$$
\boxed{\text{Find } w,b \text{ such that } J(w,b) \text{ is minimized}}
$$

## Important Concepts

To understand linear regression, we need the following concepts:

1. **Data points** — $(x_i,y_i)$
2. **Linear model** — $\hat{y}=wx+b$
3. **Prediction** — $\hat{y}_i$
4. **Residual / error** — $y_i-\hat{y}_i$
5. **Loss / cost function** — measures the total error
6. **Least squares** — minimizes the squared errors
7. **Optimization** — finds the $w,b$ that minimize the cost function
8. **Derivative / gradient** — tells us how the cost changes when $w$ or $b$ changes
9. **Gradient descent** — an iterative method for finding the minimum

## Key Idea

> **Linear regression finds the parameters of a line that minimize the overall prediction error, commonly measured as the sum of squared residuals.**

# Step 4: Understanding the Concept of Mean Square Error (MSE)

We have a line that is supposed to represent our data.
Now we want to answer: How different is this line from the actual data?

For each data point, we can look at the vertical distance between the actual point and the line.

The line gives us a predicted value: $\hat{y}=wx+b$ for x.

For a data point with actual value $y$, the difference is: $e=y−\hat{y}$​   
We call this difference the error. How our estimation different from the original data.

But we dont have one data, we have multiple. If we sum up all errors we can get a single number precisely representing the error. So we can compare more lines if the one is better than other!

But there is some problem, some errors are positive whereas others are negative, summing them will not give a good result. We eventually interested with the distance not the positive or negative. So overcoming this problem we square the errors so we always have positive outcomes. 

Finally we average the sum of errors which gives us the Mean Squared Error (MSE).

$$
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

Smaller MSE means that line is closer to overall data,
Larger MSE means that line is farther from the overal data.

Now lets practice this.





````python
# Calculate predictions for the actual data points
y_pred = w * study_hours + b

# Calculate the error for each data point
errors = exam_scores - y_pred

# Square the errors
squared_errors = errors ** 2

# Calculate the Mean Squared Error
mse = np.mean(squared_errors)

print("Errors:", errors)
print("Squared errors:", squared_errors)
print("MSE:", mse)
````

# Step 6: Systematization of process

Now we have a way to compare different candidate lines using MSE.

The question is: Can we systematically find the best possible line without manually guessing different values for $w$ and $b$? There is - as I can find - various methods to achieve this. 

We will look at them because the three methods teach three different ways of thinking about the same problem. We don't learn them because you always need all three in practice.

We will go one over another. The systematic search emthods we will inquire here are:

1. Grid search (brute force)
2. Gradient descent 
3. Normal equation (analytical solution)

# Step 6.1: Grid search

The idea of grid search as an algorithm goes in such steps:
- Define possible values for $w$ and $b$
- Try every combination
- Calculate MSE for each combination.
- Choose combination with the smallest MSE

This is easy to understand, but inefficient for many parameters. In fact we already practiced in previous step when we calculate and compare three different line versions above. So I am skipping example for this method.

# Step 6.2: Gradient descent

Gradient → tells us the direction in which the error changes most strongly.
Descent → we move in the direction that makes the error smaller.

In this method instead of trying set of $w$ and $b$ values; we go like this:

1. Start with initial values for $w$ and $b$.
2. Calculate the predictions of the line.
3. Calculate the MSE.
4. Determine how w and b should change to reduce the MSE.
5. Update $w$ und $b$.
6. Calculate the MSE again.
7. Repeat this process until the MSE becomes sufficiently small.

The idea is to gradually move toward values of w and b that produce a better-fitting line.

````python
# starting from step 4 as previous steps already iterated on this text.
# Keep the intercept b fixed so we can focus only on w
b = 30

# Create different possible values for the slope w
# np.arange(start, stop, step) creates values from 2 up to 8
# with a step size of 0.5
w_values = np.arange(2, 8, 0.5)

# Create an empty list to store the MSE for each value of w
mse_values = []

# Try each possible value of w
for w in w_values:
    # Calculate the predicted exam scores using y = wx + b
    y_pred = w * study_hours + b

    # Calculate the difference between actual and predicted scores
    errors = exam_scores - y_pred

    # Square the errors so positive and negative errors do not cancel out
    squared_errors = errors ** 2

    # Calculate the average squared error (MSE)
    mse = np.mean(squared_errors)

    # Store the MSE so we can compare it later
    mse_values.append(mse)

# Print each value of w together with its corresponding MSE
for w, mse in zip(w_values, mse_values):
    print("w =", w, "MSE =", mse)
````

````python
# Plot the relationship between w and MSE
# Each point represents one possible value of w
# and the MSE calculated for that value
plt.plot(w_values, mse_values, marker="o")

# Label the x-axis with the parameter we are changing
plt.xlabel("w")

# Label the y-axis with the error we are measuring
plt.ylabel("MSE")

# Add a title to explain what the plot shows
plt.title("How w Affects the MSE")

# Display the plot
plt.show()
````

The gradient describes rate and direction of change, which you can visualize as slope of the curve. We can clearly see that on certain $w$ value we have the minimum MSE. 

But what if we have selected value that is too big? For instance 6 then we get only bigger values, which provides us the idea that we need to go further down.

The gradient gives us this information mathematically.

Now lets do the reverse. Keep the $w$ fixed and try to find best fitting $b$ value:

````python
# Keep w fixed so we can focus only on b
w = 5

# Create different possible values for b
b_values = np.arange(10, 50, 1)

# Create an empty list to store the MSE for each value of b
mse_values = []

# Try each possible value of b
for b in b_values:
    # Calculate the predicted exam scores using y = wx + b
    y_pred = w * study_hours + b

    # Calculate the difference between actual and predicted scores
    errors = exam_scores - y_pred

    # Square the errors
    squared_errors = errors ** 2

    # Calculate the Mean Squared Error
    mse = np.mean(squared_errors)

    # Store the MSE
    mse_values.append(mse)

# Print each value of b together with its MSE
for b, mse in zip(b_values, mse_values):
    print("b =", b, "MSE =", mse)
````

````python
# Plot the relationship between b and MSE
plt.plot(b_values, mse_values, marker="o")

plt.xlabel("b")
plt.ylabel("MSE")
plt.title("How b Affects the MSE")

plt.show()
````

We can apply the same idea to $b$.

We keep $w$ fixed and try different values of $b$. For each value, we calculate the MSE.

By plotting $b$ against MSE, we get a curve with a minimum.

If we start with a $b$ value that is too small, we are on the left side of the curve. We need to increase $b$ to move toward the minimum.

If we start with a $b$ value that is too large, we are on the right side of the curve. We need to decrease $b$ to move toward the minimum.

Therefore, just like with $w$, the gradient of the MSE with respect to $b$ tells us which direction we should move.

# Step 6.3: Analytic Solution

The previous methods give us ways to search for a good solution.

With grid search, we try many possible values of $w$ and $b$ and compare their MSE.

With gradient descent, we start somewhere and gradually move toward a lower MSE.

Both methods can get us very close to the minimum, but they rely on searching and approximating the solution.

But what if we could use mathematics to find the minimum directly?

Instead of trying many values, we can describe the MSE mathematically and ask:

> For which values of $w$ and $b$ is the MSE at its minimum?

This leads us to the analytic solution. The basic idea comes from calculus: At the minimum of a smooth curve, its gradient is zero.

For our linear regression problem, the MSE depends on two parameters:

$$
\frac{1}{n}
\sum_{i=1}^{n}
\left(y_i-(wx_i+b)\right)^2
$$

We can then solve these equations mathematically to obtain the values of $w$ and $b$ that minimize the MSE.

So instead of:

```
Try → calculate MSE → compare → try again
```

we can do:

```
    I           II                      III             IV          V
Write MSE → differentiate → set derivatives to zero → solve → obtain w and b
```

This is called the analytic solution because we derive the solution directly from the mathematics of the problem.

For ordinary linear regression, this eventually leads to the normal equation, which gives the optimal parameters directly.

The important idea is not the formula yet.

We first want to understand why setting the gradient to zero allows us to find the minimum.

# Step 6.3.1: Write the MSE

````python
# STEP 1: Write the MSE
#
# We start with our current line:
# y_pred = w * x + b
#
# We use it to calculate the prediction for each data point.

y_pred = w * study_hours + b

# Calculate the error for each data point
errors = exam_scores - y_pred

# Square the errors
squared_errors = errors ** 2

# Calculate the Mean Squared Error
mse = np.mean(squared_errors)

# Print the initial MSE
print("Initial MSE:", mse)
````

## Step 6.3.2: Differentiate the MSE

In Step 1, we defined the MSE:

$$
MSE(w,b)
=
\frac{1}{n}
\sum_{i=1}^{n}
\left(y_i-(wx_i+b)\right)^2
$$

Now we want to know:

> **How does the MSE change when we change $w$ or $b$?**

To answer this, we use **derivatives**.

Because we have two parameters, $w$ and $b$, we calculate two derivatives.

### Derivative with respect to $w$

$$
\frac{\partial MSE}{\partial w}
=
-\frac{2}{n}
\sum_{i=1}^{n}
x_i
\left(y_i-(wx_i+b)\right)
$$

This tells us how the MSE changes when we change the slope $w$, while keeping $b$ fixed.

### Derivative with respect to $b$

$$
\frac{\partial MSE}{\partial b}
=
-\frac{2}{n}
\sum_{i=1}^{n}
\left(y_i-(wx_i+b)\right)
$$

This tells us how the MSE changes when we change the intercept $b$, while keeping $w$ fixed.

The derivatives are therefore the **gradients** we discussed earlier.

* If the derivative is negative, increasing the parameter decreases the MSE.
* If the derivative is positive, increasing the parameter increases the MSE.
* If the derivative is zero, the MSE is locally flat.

At this stage, we have only **calculated the derivatives**.

We have not yet found the minimum.

The next step is to ask:

> **What should these derivatives be at the minimum?**


````python
# STEP 2: Differentiate the MSE
#
# Calculate the derivative of MSE with respect to w
# and the derivative of MSE with respect to b.

n = len(study_hours)

# Derivative of MSE with respect to w
gradient_w = (-2 / n) * np.sum(
    study_hours * (exam_scores - (w * study_hours + b))
)

# Derivative of MSE with respect to b
gradient_b = (-2 / n) * np.sum(
    exam_scores - (w * study_hours + b)
)

# Print the derivatives
print("Derivative with respect to w:", gradient_w)
print("Derivative with respect to b:", gradient_b)
````

# Step 6.3.3 Set the derivates to zero

I believe it is clear that we need to set them to zero to find where the value change slope is being horizontal.

````python
# STEP 3: Set the derivatives to zero
#
# At the minimum of the MSE, the derivatives must be zero.
#
# Therefore, we want:
#
# dMSE/dw = 0
# dMSE/db = 0

print("Gradient with respect to w:", gradient_w)
print("Gradient with respect to b:", gradient_b)

print("\nAt the minimum:")
print("dMSE/dw = 0")
print("dMSE/db = 0")
````

### Step 6.3.4: Solve the equations (Solve for $b$)

We start with the derivative with respect to $b$:

$$
\frac{\partial MSE}{\partial b}
=
-\frac{2}{n}
\sum_{i=1}^{n}
\left(y_i-(wx_i+b)\right)
$$

From Step 3, at the minimum:

$$
\frac{\partial MSE}{\partial b}=0
$$

Therefore:

$$
-\frac{2}{n}
\sum_{i=1}^{n}
\left(y_i-(wx_i+b)\right)
=0
$$

The constant $-\frac{2}{n}$ does not affect the solution, so we can remove it:

$$
\sum_{i=1}^{n}
\left(y_i-(wx_i+b)\right)=0
$$

Expand the sum:

$$
\sum y_i-w\sum x_i-\sum b=0
$$

Since $b$ is the same for every data point:

$$
\sum b=nb
$$

Therefore:

$$
\sum y_i-w\sum x_i-nb=0
$$

Rearrange to isolate $b$:

$$
nb=\sum y_i-w\sum x_i
$$

Divide by $n$:

$$
b=
\frac{\sum y_i}{n}
-
w\frac{\sum x_i}{n}
$$

Using the means:

$$
\bar{x}=\frac{\sum x_i}{n}
$$

and

$$
\bar{y}=\frac{\sum y_i}{n}
$$

we obtain:

$$
\boxed{b=\bar{y}-w\bar{x}}
$$

We now have an equation for $b$ in terms of $w$.

We still need to find the optimal value of $w$.


### Step 6.3.4: Solve the equations (Solve for $w$)

We start with the derivative with respect to $w$:

$$
\frac{\partial MSE}{\partial w}
=
-\frac{2}{n}
\sum_{i=1}^{n}
x_i\left(y_i-(wx_i+b)\right)
$$

From Step 3, at the minimum:

$$
\frac{\partial MSE}{\partial w}=0
$$

Therefore:

$$
-\frac{2}{n}
\sum_{i=1}^{n}
x_i\left(y_i-(wx_i+b)\right)
=0
$$

Again, the constant $-\frac{2}{n}$ does not affect the solution:

$$
\sum_{i=1}^{n}
x_i\left(y_i-(wx_i+b)\right)=0
$$

Expand the expression:

$$
\sum x_i y_i
-
w\sum x_i^2
-
b\sum x_i
=0
$$

Rearrange:

$$
w\sum x_i^2
=
\sum x_i y_i
-
b\sum x_i
$$

Therefore:

$$
w=
\frac{
\sum x_i y_i-b\sum x_i
}{
\sum x_i^2
}
$$

But we already found:

$$
b=\bar{y}-w\bar{x}
$$

Substituting this into the equation above and simplifying gives:

$$
\boxed{
w=
\frac{
\sum_{i=1}^{n}(x_i-\bar{x})(y_i-\bar{y})
}{
\sum_{i=1}^{n}(x_i-\bar{x})^2
}
}
$$

Now we have the optimal value of $w$.

Once we know $w$, we can use our previous result:

$$
\boxed{b=\bar{y}-w\bar{x}}
$$

Therefore, the analytic solution gives us both parameters directly:

$$
\boxed{
w=
\frac{
\sum(x_i-\bar{x})(y_i-\bar{y})
}{
\sum(x_i-\bar{x})^2
}
}
$$

$$
\boxed{
b=\bar{y}-w\bar{x}
}
$$


### Step 6.3.5: Obtain values for $w$ and $b$

````python
# STEP 5: Obtain w and b
#
# We now use the formulas derived analytically
# to calculate the optimal values of w and b.

# Calculate the deviations from the mean
x_deviation = study_hours - x_mean
y_deviation = exam_scores - y_mean

# Calculate the numerator of w
numerator = np.sum(x_deviation * y_deviation)

# Calculate the denominator of w
denominator = np.sum(x_deviation ** 2)

# Calculate the optimal slope w
w_analytic = numerator / denominator

# Calculate the optimal intercept b
b_analytic = y_mean - w_analytic * x_mean

# Print the results
print("Analytic w:", w_analytic)
print("Analytic b:", b_analytic)
````

````python
# Finally we can calculate the exact line.
# Calculate predictions using the analytic solution
y_pred_analytic = w_analytic * study_hours + b_analytic

# Calculate the errors
errors_analytic = exam_scores - y_pred_analytic

# Calculate the MSE
mse_analytic = np.mean(errors_analytic ** 2)

# Print the final MSE
print("Analytic MSE:", mse_analytic)
````

````python
# Plot the original data
plt.scatter(
    study_hours,
    exam_scores,
    label="Data"
)

# Plot the analytical solution
plt.plot(
    x_line,
    y_line,
    label=f"Analytical solution: y = {w_analytic:.2f}x + {b_analytic:.2f}"
)

plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Linear Regression — Analytical Solution")
plt.legend()
plt.show()
````
