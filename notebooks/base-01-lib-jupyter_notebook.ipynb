# Anatomy of jupyter note.

1. Cell
    Type — Code / Markdown

2. Input — what you type

3. Output — what it produces (code cells only)

4. Iterator — In [n], Out[n]

5. Kernel — runs the code

Mode — Edit / Command
Run — Shift+Enter

# To run code

## VSC Extension 
On top right I see `Select Kernel` option.  I follow these steps.
- In powershell
    > Get-Command python 
- added `ipykernel` as dependency on my virtual env.
    > python -m pip install ipykernel
- register venv as kernel
    > python -m ipykernel install --user --name=ml-lab --display-name="Python (ml-lab)"  
    ```
    Installed kernelspec ml-lab in <path>
    ```
- check if its installed
    > python -m jupyter kernelspec list
    ```
    python3    <path>
    ml-lab     <path>
    ```
- ctrl shift p > developer reload window
