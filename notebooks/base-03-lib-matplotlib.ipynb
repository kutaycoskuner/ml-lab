# matplotlib - Why it exists and why we need it?

## Visualization (ai-gen)

We need to **see** the data and the model, not just read numbers.
`matplotlib` is the standard Python plotting library — it turns arrays
into charts directly inside the notebook.

**Why we need it:**
- See the shape of the data (is it linear? noisy? clustered?)
- Plot the fitted line against the data points
- Watch the loss go down during training

**Core usage:**
```python
import matplotlib.pyplot as plt

plt.scatter(x, y)          # dots
plt.plot(x, y_pred)        # line
plt.xlabel("x")
plt.ylabel("y")
plt.title("...")
plt.show()
```
