# Numpy - Why it exists and why we need it?
Python lists are slow and awkward for math. NumPy fixes both. Everything in ML — pandas, scikit-learn, PyTorch — sits on top of it.

Imagine you have 1 million numbers and want to multiply each by 2.

```py
a = list(range(1_000_000))
b = [x * 2 for x in a]   # loop over 1M items in Python
```

Each item is a full Python object (with type info, reference count, etc.)
The loop runs in Python, one item at a time
Slow — seconds for large data

```py
import numpy as np
a = np.arange(1_000_000)
b = a * 2                # one vectorized operation
```

Data stored as a compact block of raw numbers (like C arrays)

The multiplication happens in compiled C code, not Python

Fast — milliseconds

That's the whole point.


Python lists are slow, memory-heavy. Provide no vectorized-math (this is more of computer science problem rather than data-science), no linear algebra. NumPy provide a solution for this.

1. The ndarray — a fast, typed, n-dimensional array

```py
a = np.array([1, 2, 3])          # 1D
b = np.array([[1, 2], [3, 4]])   # 2D (matrix)
c = np.zeros((3, 3))             # 3x3 of zeros
```

2. Vectorization 
```py
x = np.array([1, 2, 3, 4])
x + 10        # [11, 12, 13, 14]
x * 2         # [2, 4, 6, 8]
x ** 2        # [1, 4, 9, 16]
np.sqrt(x)    # [1, 1.41, 1.73, 2]
```

3.  Broadcasting — combining different shapes
```py
a = np.array([[1, 2, 3],
              [4, 5, 6]])       # shape (2, 3)
b = np.array([10, 20, 30])      # shape (3,)
a + b                           # b is "stretched" across rows
# [[11, 22, 33],
#  [14, 25, 36]]
```

4. Linear algebra
```py
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
A @ B                # matrix multiply
np.linalg.inv(A)     # inverse
np.linalg.eig(A)     # eigenvalues
```

# NumPy is the foundation of the whole stack
```
NumPy
 ├── pandas          (labeled tables, built on NumPy arrays)
 ├── scikit-learn    (classical ML, takes NumPy arrays)
 ├── matplotlib      (plots NumPy arrays)
 ├── PyTorch         (tensors — like NumPy arrays + GPU + autograd)
 └── TensorFlow      (same idea)
 ```

For instance PyTorch tensors are basically NumPy arrays with two additions: they run on GPU, and they track gradients automatically. So NumPy knowledge transfers directly.

# When NOT to use NumPy

- Labeled/tabular data → use pandas (which uses NumPy underneath)
- Deep learning with autograd + GPU → use PyTorch/TensorFlow
- Tiny scripts where a list is fine → don't over-engineer
- Text/strings → NumPy is numeric-focused; use Python strings/lists
